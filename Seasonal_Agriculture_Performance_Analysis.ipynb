## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")
```

## 2. Load the Dataset

```python
df = pd.read_csv("seasonal_agriculture_performance_dataset.csv")
print("Dataset shape:", df.shape)
display(df.head())
```

## 3. Dataset Overview & Data Quality

```python
print("Rows:", len(df))
print("Columns:", len(df.columns))
print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
display(df.dtypes.to_frame("dtype"))

print("\nMissing values:")
display(df.isna().sum().sort_values(ascending=False).to_frame("missing_values"))

print("\nDuplicate rows:", df.duplicated().sum())
```

## 5. Seasonal Performance

```python
season_order = ["Kharif", "Rabi", "Zaid"]

season_summary = (
    df.groupby("Season")
      .agg(
          Records=("Farm_ID", "count"),
          Avg_Yield_t_ha=("Yield_Tonnes_Ha", "mean"),
          Avg_Production=("Production_Tonnes", "mean"),
          Avg_Revenue_INR=("Revenue_INR", "mean"),
          Avg_Cost_INR=("Total_Cost_INR", "mean"),
          Avg_Profit_INR=("Profit_INR", "mean"),
          Avg_Rainfall_mm=("Rainfall_mm", "mean"),
          Avg_Water_Efficiency=("Water_Efficiency_t_per_1000m3", "mean"),
          Avg_Disease_Pest_Risk_pct=("Disease_Pest_Risk_pct", "mean"),
      )
      .reindex(season_order)
)
display(season_summary)
```

### Yield by Season

```python
ax = season_summary["Avg_Yield_t_ha"].plot(kind="bar", figsize=(8,5), title="Average Yield by Season")
ax.set_xlabel("Season")
ax.set_ylabel("Yield (tonnes/ha)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### Profit by Season

```python
ax = (season_summary["Avg_Profit_INR"] / 1000).plot(kind="bar", figsize=(8,5), title="Average Profit by Season")
ax.set_xlabel("Season")
ax.set_ylabel("Average Profit (₹ thousands)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 6. Environmental & Resource Patterns

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

season_summary["Avg_Rainfall_mm"].plot(kind="bar", ax=axes[0], title="Average Rainfall by Season")
axes[0].set_xlabel("Season")
axes[0].set_ylabel("Rainfall (mm)")
axes[0].tick_params(axis="x", rotation=0)

season_summary["Avg_Water_Efficiency"].plot(kind="line", marker="o", ax=axes[1], title="Water Efficiency by Season")
axes[1].set_xlabel("Season")
axes[1].set_ylabel("t / 1000 m³")
axes[1].set_xticks(range(len(season_order)))
axes[1].set_xticklabels(season_order)

plt.tight_layout()
plt.show()
```

## 7. Crop-wise & Season-wise Analysis

```python
crop_season = (
    df.pivot_table(
        index="Crop",
        columns="Season",
        values="Yield_Tonnes_Ha",
        aggfunc="mean"
    )
    .reindex(columns=season_order)
)

display(crop_season)

ax = crop_season.plot(kind="barh", figsize=(10,6), title="Average Yield by Crop and Season")
ax.set_xlabel("Average Yield (tonnes/ha)")
ax.set_ylabel("Crop")
plt.tight_layout()
plt.show()
```

## 8. Irrigation Analysis

```python
irrigation_summary = (
    df.groupby("Irrigation_Method")
      .agg(
          Avg_Yield_t_ha=("Yield_Tonnes_Ha", "mean"),
          Avg_Profit_INR=("Profit_INR", "mean"),
          Avg_Water_Efficiency=("Water_Efficiency_t_per_1000m3", "mean")
      )
      .sort_values("Avg_Yield_t_ha", ascending=False)
)

display(irrigation_summary)

ax = irrigation_summary["Avg_Yield_t_ha"].plot(kind="bar", figsize=(8,5), title="Average Yield by Irrigation Method")
ax.set_xlabel("Irrigation Method")
ax.set_ylabel("Yield (tonnes/ha)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 9. Correlation with Profit

```python
numeric_cols = [
    "Yield_Tonnes_Ha",
    "Production_Tonnes",
    "Revenue_INR",
    "Total_Cost_INR",
    "Profit_INR",
    "Rainfall_mm",
    "Water_Efficiency_t_per_1000m3",
    "Disease_Pest_Risk_pct"
]

profit_corr = df[numeric_cols].corr(numeric_only=True)["Profit_INR"].sort_values(ascending=False)
display(profit_corr.to_frame("Correlation_with_Profit"))
```

## 10. Key Findings

- Average water-efficiency values are approximately **5.89**, **5.19**, and **4.41 t/1000 m³** for Kharif, Rabi and Zaid respectively.
- The dataset shows an average disease/pest risk of about **54.5% in Kharif**, higher than the other two seasons.
